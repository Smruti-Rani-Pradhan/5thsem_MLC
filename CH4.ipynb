```python
import pandas as pd
room_length=[18,20,10,12,18,11]
room_breadth=[29,20,10,12,19,16]
room_type=['Big','Big','Normal','Normal','Big','Normal']
data=pd.DataFrame({'Length':room_length,'Breadth':room_breadth,'Type':room_type})
print(data)
```

```python
data['Area']=data['Length']*data['Breadth']
print(data)
```

```python
import pandas as pd
age=[18,20,23,19,18,22,21]
city=['City A','City B','City B','City A','City A','City C','City B']
data=pd.DataFrame({'age':age, 'city':city})
print(data)
```

```python
dummy_features=pd.get_dummies(data['city'])
data_age=pd.DataFrame(data=data,columns=['age'])
data_mod=pd.concat([data_age.reset_index(drop=True),dummy_features],axis=1)
print(data_mod)
```

```python
import pandas as pd
from sklearn import preprocessing

marks_science=[78,56,87,91,45,62]
marks_maths=[75,78,56,87,91,45]
grade=['B','C','A','A','D','B']
data=pd.DataFrame({'Science marks':marks_science,
                   'Math marks': marks_maths,
                   'Total grade': grade 
                  })
print(data)
```

```python
#option 1- use if dataset islarge
le=preprocessing.LabelEncoder()
le.fit(data['Total grade'])
data['Total grade']=le.transform(
data['Total grade'])
print(data)
```

```python
#option 2
target=data["Total grade"].replace(['A','B','C'])
predictors=data.iloc[:,0:2]
data_mod=pd.concat([predictors.reset_index(drop=True),target],axis=1)
print(data_mod)
```
